```python
import pandas as pd
from pathlib import Path

# Carpeta donde están los archivos
carpeta = Path("../data")

# Archivos a concatenar, en orden
archivos = [carpeta / f"BD {año}.xlsx" for año in range(2020, 2026)]

# Obtener las hojas de cada archivo
hojas_por_archivo = {}

for archivo in archivos:
    excel = pd.ExcelFile(archivo)
    hojas_por_archivo[archivo.name] = excel.sheet_names
    
    print(f"\n{archivo.name}:")
    print(f"  Número de hojas: {len(excel.sheet_names)}")
    print(f"  Hojas: {excel.sheet_names}")


# Obtener todas las hojas existentes en cualquiera de los archivos
hojas = []

for archivo in archivos:
    for hoja in hojas_por_archivo[archivo.name]:
        if hoja not in hojas:
            hojas.append(hoja)

print("\n========================================")
print("Todas las hojas encontradas:")
print(f"Total: {len(hojas)}")
print(hojas)
print("========================================\n")


# Diccionario donde guardaremos los DataFrames finales
datos_concatenados = {}


# Recorrer cada hoja
for hoja in hojas:

    dfs = []

    print(f"\nProcesando hoja: {hoja}")

    # Recorrer los archivos en orden: 2020, 2021, ..., 2025
    for archivo in archivos:

        # Verificar si esta hoja existe en el archivo
        if hoja in hojas_por_archivo[archivo.name]:

            print(f"  Leyendo: {archivo.name}")

            df = pd.read_excel(
                archivo,
                sheet_name=hoja
            )

            dfs.append(df)

        else:
            print(f"  OMITIDA: {archivo.name} no tiene la hoja '{hoja}'")


    # Concatenar únicamente los archivos que tienen esta hoja
    if dfs:
        datos_concatenados[hoja] = pd.concat(
            dfs,
            ignore_index=True
        )


# Guardar todo en un nuevo Excel
archivo_salida = carpeta / "BD_2020_2025_concatenada.xlsx"

with pd.ExcelWriter(
    archivo_salida,
    engine="openpyxl"
) as writer:

    for hoja, df in datos_concatenados.items():

        df.to_excel(
            writer,
            sheet_name=hoja,
            index=False
        )


print("\n========================================")
print("¡LISTO!")
print(f"Archivo generado: {archivo_salida}")
print(f"Total de hojas generadas: {len(datos_concatenados)}")
print("========================================")
```
